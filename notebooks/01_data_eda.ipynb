## Temperature conversion

The zip Readme and the UCI web page give different formulas for converting temp back to degrees Celsius. We compared both against the observed daily mean temperature at Reagan National Airport (KDCA), estimated as (max + min) / 2.

| Date | KDCA observed mean (°C) | Formula A: temp × 41 | Formula B: temp × 47 − 8 |
|---|---|---|---|
| 2011-01-22 | ≈ −5 | 2.4 | −5.2 |
| 2011-04-15 | ≈ 13.5 | 18.3 | 13.0 |
| 2011-07-22 | ≈ 33.5 | 34.8 | 31.9 |
| 2012-01-04 | ≈ −3 | 4.4 | −2.9 |

**Decision:** The zip Readme and the UCI web page give different formulas. On the two winter days (2011-01-22 and 2012-01-04), Formula A gives temperatures above zero, which contradicts the observed sub-zero means, whereas Formula B is within 1°C. Formula B is also closer on 2011-04-15, and the two are similar on 2011-07-22, so we use Formula B.
